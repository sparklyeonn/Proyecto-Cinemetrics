1. Configuración e Importación

```python
import sys
import os
import pandas as pd

# 1. Para encontrar src/utils.py
sys.path.append(os.path.abspath('../src'))

# 2. Importar funciones desde el archivo modular
from utils import cargar_y_unir_datasets, extraer_director

print(" Funciones cargadas desde el módulo src.utils")
```

2. Unión de Tablas (Merge) : Combinando la información de las películas con la de los créditos (actores, directores, etc.).

```python
# Defino rutas a los archivos raw
ruta_movies = '../data/raw/tmdb_5000_movies.csv'
ruta_credits = '../data/raw/tmdb_5000_credits.csv'

# 3. UNIÓN DE TABLAS: Combinar por la columna 'id'
df_proyecto = cargar_y_unir_datasets(ruta_movies, ruta_credits)

print(f" Unión completada. El dataset ahora tiene {df_proyecto.shape[1]} columnas.")
```

3. Estructuración y Limpieza Inicial : Utilizo función para extraer al director y use el hallazgo de Conny (el tema de los presupuestos en $0) para dejar los datos listos.

```python
# 4. ESTRUCTURACIÓN: Extraer el director
df_proyecto['director'] = df_proyecto['crew'].apply(extraer_director)

# 5. FILTRADO: Limpiar "ceros" según el reporte de Conny
#  películas que tienen presupuesto y ganancias reales
df_analisis = df_proyecto[(df_proyecto['budget'] > 0) & (df_proyecto['revenue'] > 0)].copy()

print(" Datos estructurados y filtrados.")
```

4. Agrupaciones Base (Groupby) : Segmentar los datos para empezar a ver patrones. 

Por Director: Agrupando por director, el presupuesto medio dice si un director suele manejar grandes superproducciones (como James Cameron) o películas independientes de bajo costo.

Por Idioma: Agrupando por original_language, el presupuesto medio de las películas en inglés suele ser mucho más alto que el de películas en Telugu - India, lo que refleja la diferencia de poder económico de las industrias.

El presupuesto medio nos ayuda a decir: "En promedio, una película de este tipo cuesta X cantidad de dinero".

```python
# 6. AGRUPACION: ¿Qué directores tienen mayor presupuesto promedio?
top_directores_presupuesto = df_analisis.groupby('director')['budget'].mean().sort_values(ascending=False).head(10)

# 7. AGRUPACION: ¿Cuál es el ingreso promedio por idioma original?
ingresos_por_idioma = df_analisis.groupby('original_language')['revenue'].mean().sort_values(ascending=False).head(5)

print("--- Top 10 Directores por Presupuesto Medio ---")
print(top_directores_presupuesto)

print("\n--- Top 5 Idiomas por Ingresos Medios ---")
print(ingresos_por_idioma)
```
