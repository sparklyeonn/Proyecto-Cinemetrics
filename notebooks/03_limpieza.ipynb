### **Parte 3: Limpieza y sanitización de datos**

```python
import pandas as pd
import sys
import os

# Configuración de rutas y carga de funciones personalizadas
sys.path.append(os.path.abspath('../src'))

try:
    from utils import auditoria_integridad, sanear_datos
    print("Funciones de utilidad cargadas correctamente.")
except ImportError:
    print("Error: No se pudo encontrar 'utils.py' en la carpeta 'src'.")

# Carga del dataset original (Raw Data)
ruta_entrada = '../data/raw/tmdb_5000_movies.csv'

if os.path.exists(ruta_entrada):
    df_raw = pd.read_csv(ruta_entrada)
    print(f"Dataset cargado: {df_raw.shape[0]} filas y {df_raw.shape[1]} columnas.")
else:
    print(f"Error: El archivo no existe en {ruta_entrada}")

# Filtrado y limpieza de valores en cero
antes = len(df_raw)

# Filtro: solo con películas que tienen datos financieros reales
df_filtrado = df_raw[(df_raw['budget'] > 0) & (df_raw['revenue'] > 0)].copy()

despues = len(df_filtrado)
eliminados = antes - despues

print(f"Limpieza financiera completada:")
print(f"- Registros eliminados (Presupuesto/Ingreso = 0): {eliminados}")
print(f"- Registros restantes para análisis: {despues} ({(despues/antes)*100:.1f}%)")

# Limpieza de datos (con función externa): de nulos y formatos definidos en utils
df_final = sanear_datos(df_filtrado)

# 5. Auditoría de Integridad
# Verificamos que el dataset final cumpla con los estándares mínimos
calidad_ok = auditoria_integridad(df_final)

if calidad_ok:
    print("\nLos datos han pasado la auditoría de calidad y están listos")
else:
    print("\nSe detectaron inconsistencias menores")

# Guardar el dataset procesado
ruta_salida = '../data/processed/movies_unificado.csv'

os.makedirs('../data/processed', exist_ok=True)

df_final.to_csv(ruta_salida, index=False)

print("-" * 30)
print(f"Archivo generado: {ruta_salida}")
```
