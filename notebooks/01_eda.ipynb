### **Carga de Dataset + Ver Columnas**

```python
import kagglehub
import os
import shutil
import matplotlib.pyplot as plt
import seaborn as sns

# Descargar
path = kagglehub.dataset_download("tmdb/tmdb-movie-metadata")
print(f"Directorio original de Kaggle: {path}")

# Ver qué archivos hay dentro de la carpeta oculta
archivos_descargados = os.listdir(path)
print(f"Archivos encontrados: {archivos_descargados}")

# Moverlos a la carpeta raw
destino = "../data/raw"
os.makedirs(destino, exist_ok=True) 

for archivo in archivos_descargados:
    ruta_origen = os.path.join(path, archivo)
    ruta_destino = os.path.join(destino, archivo)
    shutil.copy(ruta_origen, ruta_destino)
    print(f"Copiado {archivo} a {destino}")
```

```python
import pandas as pd
import os

# Definir las rutas
movies_path = '../data/raw/tmdb_5000_movies.csv'
credits_path = '../data/raw/tmdb_5000_credits.csv'

# Verificar si los archivos existen 
if os.path.exists(movies_path) and os.path.exists(credits_path):
    print("Archivos encontrados en /data/raw")
    
    # Cargar los datasets
    df_movies = pd.read_csv(movies_path)
    df_credits = pd.read_csv(credits_path)
    
    print("Datasets cargados exitosamente")
else:
    print("Error: No se encontraron los archivos. Revisa la carpeta /data/raw")
```

```python
# Ver las primeras filas
print("Primeras filas de Movies")
display(df_movies.head())

# Ver la estructura (columnas y tipos de datos)
print("\nInformación del Dataset")
df_movies.info()
columnas_financieras = ['budget', 'revenue']
conteo_ceros = (df_movies[columnas_financieras] == 0).sum()

plt.figure(figsize=(8, 5))
sns.barplot(x=conteo_ceros.index, y=conteo_ceros.values, palette='magma')
plt.title('Detección de "Nulos Disfrazados" (Valores en $0)')
plt.ylabel('Cantidad de Registros')
for i, v in enumerate(conteo_ceros.values):
    plt.text(i, v + 50, str(v), ha='center', fontweight='bold')
plt.show()

# Ver si hay datos nulos (NaN)
print("\nDatos faltantes")
print(df_movies.isnull().sum())
```

#### **2. Top 10 Géneros de Películas**

```python
import ast

# Función para extraer solo los nombres de los géneros
def extraer_generos(columna):
    lista_generos = []
    for fila in columna:
        # Convertir el texto JSON a una lista de Python
        datos = ast.literal_eval(fila)
        # Sacar solo el 'name' de cada diccionario
        for item in datos:
            lista_generos.append(item['name'])
    return lista_generos

# Se aplica la función a la columna genres
todos_los_generos = extraer_generos(df_movies['genres'])

# Convertir a una Serie de Pandas para contar fácil
conteo_generos = pd.Series(todos_los_generos).value_counts()
```

```python
plt.figure(figsize=(12, 8))

# Grafico del conteo de géneros
conteo_generos.plot(kind='bar', color='skyblue')

plt.title('Frecuencia de Géneros Cinematográficos', fontsize=15)
plt.xlabel('Género', fontsize=12)
plt.ylabel('Cantidad de Películas', fontsize=12)
plt.xticks(rotation=45) 
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.savefig('../outputs/figures/distribucion_generos.png', bbox_inches='tight')
plt.show()
```
